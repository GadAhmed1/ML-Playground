```python
import pandas as pd


df = pd.read_csv(
    r'D:\Main\projects\ML-playground-\02_Manually_creating_algorithm\01_Niave_Bayes\data.csv',
    engine='pyarrow'
)
df.head()
```

```python
(
    df.groupby('play')
    .agg(
        counts=('play','count')
    )
    .reset_index()
    .to_dict()
)
```

```python
Total_Postives = df[df['play'] == 'Yes']['play'].count()
Total_Nagtives = df[df['play'] == 'No']['play'].count()
```

```python
for col in df.iloc[:,1:-1].columns:
    result = df.groupby(col).agg(
        Yes=('play', lambda x: (x == 'Yes').sum()),
        No=('play', lambda x: (x == 'No').sum())
    ).reset_index()
    result['TotalYes']=Total_Postives
    result['TotalNo']=Total_Nagtives
    result['p_Yes']=result['Yes'] / result['TotalYes']
    result['p_No']=result['No'] / result['TotalNo']
    print(result)
```

```python
result = df.groupby('outlook').agg(
    Yes=('play', lambda x: (x == 'Yes').sum()),
    No=('play', lambda x: (x == 'No').sum())
).reset_index()
result['TotalYes']=Total_Postives
result['TotalNo']=Total_Nagtives
result['p_Yes']=result['Yes'] / result['TotalYes']
result['p_No']=result['No'] / result['TotalNo']
result
```
