```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import RandomOverSampler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import classification_report
columns = [
    'fLength', 'fWidth', 'fSize', 'fConc', 'fConc1',
    'fAsym', 'fM3Long', 'fM3Trans', 'fAlpha', 'fDist', 'class'
]

df = pd.read_csv(
    r"D:\Main\projects\ML-playground-\01_beginning\magic04.data",
    engine='pyarrow',
    dtype_backend='pyarrow',
    names=columns
)
```

```python
train, valid, test = np.split(df.sample(frac=1),[int(0.6 * len(df)),int(0.8 * len(df))])
```

```python
df['class_name'] = df['class']
```

```python
df['class'] = np.where(df['class'] == 'g',1,0)
```

```python
df.sample(10)
```

```python
def scale_dataset(dataset,oversample=False):
    x = dataset.iloc[:,:-2].values
    y = dataset.loc[:,'class'].values

    scaler = StandardScaler()
    x = scaler.fit_transform(x)

    if oversample:
        ros = RandomOverSampler()
        x,y = ros.fit_resample(x,y)

    data = np.hstack((x,np.reshape(y,(-1,1))))

    return data, x, y
```

```python
train, x_train, y_train = scale_dataset(train,True)
valid, x_valid, y_valid = scale_dataset(valid,False)
test, x_test, y_test = scale_dataset(test,False)
```

```python
knn_model = KNeighborsClassifier(n_neighbors=3)
knn_model.fit(x_train,y_train)

y_predict = knn_model.predict(x_test)
```

```python
print(classification_report(y_predict,y_test))
```
